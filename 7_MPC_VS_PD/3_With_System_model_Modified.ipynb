```python
def plot(store_times, store_accelerations, store_velocities, store_seperations, store_detected_times, store_start_reaction_times, Dr, Vp, title):
    fig, axes = plt.subplots(1, 3, figsize=(90, 25))
    font = 50
    border_thickness = 8.0
    
    for i in range(3):
        for spine in axes[i].spines.values():
            spine.set_linewidth(border_thickness)

    sri_1 = store_times[0].index(store_start_reaction_times[0])
    sri_2 = store_times[1].index(store_start_reaction_times[1])

    # Separation plot
    line_pd_sep, = axes[0].plot(store_times[0], store_seperations[0], color='red', linewidth=6)
    line_mpc_sep, = axes[0].plot(store_times[1], store_seperations[1], color='blue', linestyle='-.', linewidth=6)
    axes[0].axvline(x=store_detected_times[0], color='black', linestyle='-', linewidth=6)
    axes[0].axvline(x=store_start_reaction_times[0], color="#4E13D8EB", linestyle='-', linewidth=6)
    axes[0].axvline(x=store_detected_times[1], color="#EA00FFFF", linestyle='--', linewidth=4)
    axes[0].axvline(x=store_start_reaction_times[1], color="#00FFEAFF", linestyle='-.', linewidth=4)
    axes[0].plot(store_times[0][sri_1:], [CP.Dd + CP.Td * V * 5 / 18 for V in store_velocities[0][sri_1:]], label='Desired final separation for PDC', color='green', linewidth=6)
    axes[0].plot(store_times[1][sri_2:], [CP.Dd + CP.Td * V * 5 / 18 for V in store_velocities[1][sri_2:]], label='Desired final separation for MPC', color="#965102CC", linewidth=8, linestyle=':')
    axes[0].grid(True, linewidth=4.0)
    axes[0].set_xlabel('Time [s]', fontsize=font, fontweight='bold')
    axes[0].set_ylabel('Separation [m]', fontsize=font, fontweight='bold')
    axes[0].tick_params(axis='both', which='major', labelsize=font, width=2)
    axes[0].legend(fontsize=font)

    # Velocity plot
    line_pd_vel, = axes[1].plot(store_times[0], store_velocities[0], color='red', linewidth=6)
    line_mpc_vel, = axes[1].plot(store_times[1], store_velocities[1], color='blue', linestyle='-.', linewidth=6)
    axes[1].axvline(x=store_detected_times[0], color='black', linestyle='-', linewidth=6)
    axes[1].axvline(x=store_start_reaction_times[0], color="#4E13D8EB", linestyle='-', linewidth=6)
    axes[1].axvline(x=store_detected_times[1], color="#EA00FFFF", linestyle='--', linewidth=4)
    axes[1].axvline(x=store_start_reaction_times[1], color="#00FFEAFF", linestyle='-.', linewidth=4)
    axes[1].plot(store_times[0][sri_1:], [Vp * 18 / 5 for _ in range(len(store_velocities[0]))][sri_1:], label='Desired final velocity for PDC', color='green', linewidth=6)
    axes[1].plot(store_times[1][sri_2:], [Vp * 18 / 5 for _ in range(len(store_velocities[1]))][sri_2:], label='Desired final velocity for MPC', color="#965102CC", linewidth=8, linestyle=':')
    axes[1].grid(True, linewidth=4.0)
    axes[1].set_xlabel('Time [s]', fontsize=font, fontweight='bold')
    axes[1].set_ylabel('Velocity [km/hr]', fontsize=font, fontweight='bold')
    axes[1].tick_params(axis='both', which='major', labelsize=font, width=2)
    axes[1].legend(fontsize=font)

    # Acceleration plot
    line_pd_acc, = axes[2].plot(store_times[0], store_accelerations[0], color='red', linewidth=6)
    line_mpc_acc, = axes[2].plot(store_times[1], store_accelerations[1], color='blue', linestyle='-.', linewidth=6)
    axes[2].axvline(x=store_detected_times[0], color='black', linestyle='-', linewidth=6)
    axes[2].axvline(x=store_start_reaction_times[0], color="#4E13D8EB", linestyle='-', linewidth=6)
    axes[2].axvline(x=store_detected_times[1], color="#EA00FFFF", linestyle='--', linewidth=4)
    axes[2].axvline(x=store_start_reaction_times[1], color="#00FFEAFF", linestyle='-.', linewidth=4)
    axes[2].grid(True, linewidth=4.0)
    axes[2].set_xlabel('Time [s]', fontsize=font, fontweight='bold')
    axes[2].set_ylabel('Acceleration [m/s^2]', fontsize=font, fontweight='bold')
    axes[2].tick_params(axis='both', which='major', labelsize=font, width=2)

    # Create common legend at the bottom
    common_lines = [
        line_pd_sep, 
        line_mpc_sep,
        matplotlib.lines.Line2D([0], [0], color='black', linestyle='-', linewidth=6),
        matplotlib.lines.Line2D([0], [0], color='#4E13D8EB', linestyle='-', linewidth=6),
        matplotlib.lines.Line2D([0], [0], color='#EA00FFFF', linestyle='--', linewidth=4),
        matplotlib.lines.Line2D([0], [0], color='#00FFEAFF', linestyle='-.', linewidth=4)
    ]
    common_labels = [
        'PD_Controller', 
        'MPC_Controller',
        'Obstacle detected (PDC)', 
        'Vehicle responds (PDC)', 
        'Obstacle detected (MPC)', 
        'Vehicle responds (MPC)'
    ]
    
    fig.legend(common_lines, common_labels, loc='lower center', ncol=6, fontsize=font, 
               bbox_to_anchor=(0.5, -0.08), frameon=True)
    plt.subplots_adjust(bottom=0.05)
    plt.show()
```
