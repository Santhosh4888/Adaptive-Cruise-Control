
# ODOMENU Post-Processing Notebook

This notebook:

1. Reads `odomenu.csv`
2. Plots:
   - Linear velocity in X
   - Linear velocity in Y
   - Acceleration in X
   - Acceleration in Y
3. Computes:
   - Speed magnitude
   - Yaw angle (from quaternion)
   - Yaw rate
   - Acceleration magnitude


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python

# Load CSV (skip first row if it contains topic name)
df = pd.read_csv("odomenu.csv", skiprows=1)

df.head()
```

```python

plt.figure()
plt.plot(df["time"], df["vel_x"])
plt.xlabel("Time (s)")
plt.ylabel("Linear Velocity X (m/s)")
plt.title("Linear Velocity in X")
plt.grid()
plt.show()
```

```python

plt.figure()
plt.plot(df["time"], df["vel_y"])
plt.xlabel("Time (s)")
plt.ylabel("Linear Velocity Y (m/s)")
plt.title("Linear Velocity in Y")
plt.grid()
plt.show()
```

```python

plt.figure()
plt.plot(df["time"], df["acc_x"])
plt.xlabel("Time (s)")
plt.ylabel("Acceleration X (m/s²)")
plt.title("Acceleration in X")
plt.grid()
plt.show()
```

```python

plt.figure()
plt.plot(df["time"], df["acc_y"])
plt.xlabel("Time (s)")
plt.ylabel("Acceleration Y (m/s²)")
plt.title("Acceleration in Y")
plt.grid()
plt.show()
```

```python

# Speed magnitude
df["speed_magnitude"] = np.sqrt(df["vel_x"]**2 + df["vel_y"]**2)

# Yaw angle from quaternion
qw = df["qw"]
qx = df["qx"]
qy = df["qy"]
qz = df["qz"]

df["yaw"] = np.arctan2(
    2 * (qw * qz + qx * qy),
    1 - 2 * (qy**2 + qz**2)
)

# Yaw rate (using angular velocity z)
df["yaw_rate"] = df["ang_vel_z"]

# Acceleration magnitude
df["acc_magnitude"] = np.sqrt(df["acc_x"]**2 + df["acc_y"]**2)

df[["speed_magnitude", "yaw", "yaw_rate", "acc_magnitude"]].head()
```
